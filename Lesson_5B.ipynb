## Exercise 2 - Payroll System

```python
employee_name = "David Musa"
department = "Finance"
basic_salary = 185000
bonus = 35000
tax_rate = 7.5 / 100

gross_salary = basic_salary + bonus
tax_amount = gross_salary * tax_rate
net_salary = gross_salary - tax_amount

report = f"""
================================
        PAYROLL SUMMARY
================================
Employee: {employee_name}
Department: {department}
Basic Salary: N{basic_salary:,.2f}
Bonus: N{bonus:,.2f}
Gross Salary: N{gross_salary:,.2f}
Tax(7.5%): N{tax_amount:,.2f}
Net Salary: N{net_salary:,.2f}
=================================
"""
print(report)
```

## Exercise 3 - E-commerce Order Receipt

```python
customer_name = "Amina Bello"
product = "Wireless Headphones"
quantity = 3
unit_price = 27500

total_cost = unit_price * quantity

report = f"""
===============================
     ONLINE STORE RECEIPT
===============================

Customer Name: {customer_name}
Product: {product}
Quantity: {quantity}
Unit Price: N{unit_price:,.2f}
Total Cost: N{total_cost:,.2f}

Thank you for shopping with us!
===============================
"""

print(report)
```

```python
# Challenge
customer_name = "Amina Bello"
product = "Wireless Headphones"
quantity = 3
unit_price = 27500

total_cost = unit_price * quantity
discount = total_cost * 0.10
final_amount = total_cost - discount

report = f"""
===============================
     ONLINE STORE RECEIPT
===============================

Customer Name: {customer_name}
Product: {product}
Quantity: {quantity}
Unit Price: N{unit_price:,.2f}
Total Cost: N{total_cost:,.2f}
Discount: N{discount:,.2f}
Final Amount: N{final_amount:,.2f}

Thank you for shopping with us!
===============================
"""

print(report)
```
