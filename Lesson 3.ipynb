# Question 2: Student Performance Evaluation

```python
python = 72
sql = 65
excel = 80
power_bl = 75
number = 4

total_score = python + sql + excel + power_bl
average_score = total_score / number

print("Average Score: ", average_score)
print("Passed: ", average_score >= 70)
```

# Question 3: Inventory Management System

```python
initial_stock = 250
sold_stock = 85

remaining_stock = initial_stock - sold_stock

print("Remaining Stock: ", remaining_stock)
print("Reorder Needed: ", remaining_stock < 200)
```

# Question 4: Banking Withdrawal Checker

```python
account_balance = 520000
withdrawal_amount = 150000

balance = account_balance - withdrawal_amount

print("Remaining Balance; ", balance)
print("Withdrawal Approved: ", balance >= withdrawal_amount)
```

# Bonus Challenge: Smart Shopping Discount Calculator

```python
laptop_bag = 18500
wireless_mouse = 12000
keyboard = 24500
usb_drive = 5000

total_purchase = laptop_bag + wireless_mouse + keyboard + usb_drive

print("Total Purchase: ", total_purchase)

print("Discount Qualified: ", total_purchase >= 50000)

discount = int(total_purchase * 0.1)

print("Discount Amount: ", discount)

final_amount = total_purchase - discount

print("Final Amount: ", final_amount)
```
