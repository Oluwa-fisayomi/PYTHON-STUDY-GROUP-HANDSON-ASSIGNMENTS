# Exercise 1
## Industry Scenario - Patient Registration System

```python
first_name = "Grace"
last_name = "Johnson"
age = "34 years"
blood_group = "O+"
department = "Cardiology"

print("============ CAREPLUS HOSPITAL ===========")
print("Patient Name: " + first_name + " " + last_name)
print("Age: " + age)
print("Blood Group: " + blood_group)
print("Department: " + department)
print("==========================================")
```

## Bonus Challenge 1 - Using .format() method

```python
first_name = "Grace"
last_name = "Johnson"
age = "34 years"
blood_group = "O+"
department = "Cardiology"

print("============ CAREPLUS HOSPITAL ===========")
print("Patient Name: {} {}".format(first_name, last_name))
print("Age: {}".format(age))
print("Blood Group: {}".format(blood_group))
print("Department: {}".format(department))
print("==========================================")
```

## Bonus Challenge 2 - Using f-strings

```python
first_name = "Grace"
last_name = "Johnson"
age = "34 years"
blood_group = "O+"
department = "Cardiology"

print(f"""
============ CAREPLUS HOSPITAL ===========
Patient Name: {first_name} {last_name}
Age: {age}
Blood Group: {blood_group}
Department: {department}
==========================================
""")
```

## Bonus Challenge 3 - Converting the patient's full name to uppercase

```python
first_name = ("Grace").upper()
last_name = ("Johnson").upper()
age = "34 years"
blood_group = "O+"
department = "Cardiology"

print(f"""
============ CAREPLUS HOSPITAL ===========
Patient Name: {first_name} {last_name}
Age: {age}
Blood Group: {blood_group}
Department: {department}
==========================================
""")
```

# Exercise 2

## Industry Scenario - Fraud Detection Log Analyzer

```python
bank_log = "   LOGIN SUCCESSFUL FROM ATM-045 ON 2026-07-25   "

cleaned_log = bank_log.strip()

updated_log = cleaned_log.replace("SUCCESSFUL", "VERIFIED")

print(f"""
Original Log:
{bank_log}

Cleaned Log:
{cleaned_log}

Log in Lowercase:
{cleaned_log.lower()}

Position of SUCCESSFUL:
{cleaned_log.find("SUCCESSFUL")}

Updated Log:
{updated_log}

""")
```

## Bonus Challenge

```python
print(f"""
Number of characters before cleaning:
{len(bank_log)}

Number of characters after cleaning:
{len(cleaned_log)}

Number of words:
{len(cleaned_log.split())}
""")
```
