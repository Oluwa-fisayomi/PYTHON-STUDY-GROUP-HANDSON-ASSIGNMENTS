# Exercise 3 - Inventory Management System

```python
# Task
stock = 150
stock -= 28
stock += 75
cartons = stock // 12

print("The remaining stock is: ", stock)
print("The number of full cartons: ", cartons)
```

```python
# Challenge
initial_stock = int(input("Enter the initial stock: "))
sold_items = int(input("How many items were sold? "))
new_items = int(input("How many new items arrived? "))
carton = int(input("How many items fit in one carton? "))

initial_stock -= sold_items
initial_stock += new_items
carton = initial_stock // 12
loose_items = initial_stock % carton

print("The remaining stock is: ", initial_stock)
print("The number of full cartons: ", carton)
print("The remaining loose items: ", loose_items)
```
